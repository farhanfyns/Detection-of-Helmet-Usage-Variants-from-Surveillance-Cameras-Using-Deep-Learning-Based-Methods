# **EPOCH 50 BATCH 16**

```python
from ultralytics import YOLO
import os

# Path utama proyek
base_path = r"C:\Users\fatur\Documents\Source Code TA_Helmet Detection_Farhan Faturahman"

# Path ke dataset (data.yaml)
data_yaml_path = os.path.join(base_path, "Dataset", "data.yaml")

# Path untuk menyimpan hasil training
save_dir = os.path.join(base_path, "Yolov8", "Hasil Pengujian Epoch", "Epoch 50", "Batch 16")
os.makedirs(save_dir, exist_ok=True)

# Inisialisasi model YOLOv8n (Nano)
model = YOLO("yolov8n.pt")

# Melatih model dengan dataset helm
model.train(
    data=data_yaml_path,
    epochs=50,
    imgsz=640,
    batch=16,
    lr0=0.01,
    optimizer="AdamW",
    weight_decay=0.0005,
    save_dir=save_dir
)

print(f"Model dan hasil pelatihan disimpan di: {save_dir}")
```

```python
from ultralytics import YOLO
import os
from IPython.display import Image, display
import pandas as pd

# Path ke model terbaik hasil training
model_path = r"C:\Users\fatur\Documents\Source Code TA_Helmet Detection_Farhan Faturahman\Yolov8\Hasil Pengujian Epoch\Epoch 50\Batch 16\train\weights\best.pt"

# Path ke dataset (data.yaml)
data_yaml_path = r"C:\Users\fatur\Documents\Source Code TA_Helmet Detection_Farhan Faturahman\Dataset\data.yaml"

# Path penyimpanan hasil pengujian
test_save_dir = os.path.join(os.path.dirname(save_dir))
os.makedirs(test_save_dir, exist_ok=True)

# Load model
model = YOLO(model_path)

# Evaluasi model pada dataset testing
metrics = model.val(data=data_yaml_path, split='test', save_dir=test_save_dir)
```

```python
# Path folder hasil testing
base_path = os.path.join(test_save_dir, "Batch 16", "val")

# Path file evaluasi
confusion_matrix_path = os.path.join(base_path, "confusion_matrix.png")
confusion_matrix_normalized_path = os.path.join(base_path, "confusion_matrix_normalized.png")
f1_curve_path = os.path.join(base_path, "F1_curve.png")

# Menampilkan Confusion Matrix
print("Confusion Matrix (Testing):")
display(Image(filename=confusion_matrix_path, width=600, height=600))

# Menampilkan Confusion Matrix Normalized
print("Confusion Matrix Normalized (Testing):")
display(Image(filename=confusion_matrix_normalized_path, width=600, height=600))

# Menampilkan F1 Confidence Curve
print("F1 Confidence Curve (Testing):")
display(Image(filename=f1_curve_path, width=800, height=400))

# Membaca hasil evaluasi dari CSV
results_csv_path = os.path.join(save_dir, "train", "results.csv")
results_df = pd.read_csv(results_csv_path)

print("\nTraining Results (CSV):")
pd.set_option('display.max_rows', None)  # Menampilkan semua baris
pd.set_option('display.max_columns', None)  # Menampilkan semua kolom
display(results_df)
```
