```python
import os
import glob
import pandas as pd

# Path ke folder label YOLO
label_folder = r"C:\Users\USER\Downloads\isi file txt label semua"

# Resolusi gambar (sesuai dataset)
img_width = 640
img_height = 640

# Inisialisasi list untuk menyimpan bounding box (file, width, height, area)
bounding_boxes = []

# Ambil semua file label (*.txt)
label_files = glob.glob(os.path.join(label_folder, "*.txt"))

# Loop melalui semua file label
for file in label_files:
    with open(file, "r") as f:
        lines = f.readlines()
        for line in lines:
            parts = line.strip().split()
            if len(parts) == 5:  # Format YOLO harus memiliki 5 elemen
                class_id, x_center, y_center, width, height = map(float, parts)
                width_px = int(width * img_width)  # Konversi ke pixel
                height_px = int(height * img_height)
                area_px = width_px * height_px  # Hitung luas bounding box

                if width_px > 0 and height_px > 0:  # Pastikan tidak ada nilai nol
                    bounding_boxes.append((file, width_px, height_px, area_px))

# Temukan bounding box terkecil berdasarkan luas
if bounding_boxes:
    min_bbox_entry = min(bounding_boxes, key=lambda x: x[3])  # Cari berdasarkan luas terkecil

    # Tampilkan hasil dalam pixel
    print("Bounding Box Terkecil dalam Pixel:")
    print(f"File: {min_bbox_entry[0]}")
    print(f"Width = {min_bbox_entry[1]} px, Height = {min_bbox_entry[2]} px")
    print(f"Luas Bounding Box = {min_bbox_entry[3]} px²")

    # Simpan hasil ke file CSV
    output_csv = os.path.join(label_folder, "bounding_box_sizes.csv")
    df = pd.DataFrame(bounding_boxes, columns=["File", "Width (px)", "Height (px)", "Area (px²)"])
    df.to_csv(output_csv, index=False)

    print(f"Semua ukuran bounding box telah disimpan di: {output_csv}")
else:
    print("Tidak ada bounding box valid ditemukan.")
```
