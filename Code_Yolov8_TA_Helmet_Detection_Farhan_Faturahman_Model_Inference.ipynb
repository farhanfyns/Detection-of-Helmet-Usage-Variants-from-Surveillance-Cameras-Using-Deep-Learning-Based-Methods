```python
import os
from ultralytics import YOLO

# Path ke model hasil training terbaik
model_path = r"C:\Users\USER\Documents\Source Code TA_Helmet Detection_Farhan Faturahman\Yolov8\Hasil Pengujian Epoch\Epoch 100\Batch 32\train\weights\best.pt"

# Path ke video yang akan diuji
video_path = r"C:\Users\USER\Videos\VideoProc Converter AI\Sudah Di label\MVI_2838\MVI_2838 - 003.mp4"

# Load model YOLO
model = YOLO(model_path)

if os.path.exists(video_path):
    print(f"Melakukan inference pada video: {video_path}")

    # Path untuk menyimpan hasil inference
    inference_save_dir = os.path.join(os.path.dirname(video_path), "inference_results")
    os.makedirs(inference_save_dir, exist_ok=True)
    output_video_path = os.path.join(inference_save_dir, "output.mp4")

    # Jalankan inference pada video
    model.predict(source=video_path, save=True, save_txt=True, project=inference_save_dir, name="inference_output")

    print(f"Inference selesai! Hasil disimpan di: {inference_save_dir}")
else:
    print("Video tidak ditemukan. Pastikan path sudah benar.")
```
